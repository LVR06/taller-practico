```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#1. Importacion de los datos
df = pd.read_csv("products.csv")
```

```python
#2. Verificar los datos
df.info()
df.head(10)
```

```python
#3. Verificar si hay valores nulos
print('Valores nulos por columna:')
print(df.isna().sum())
```

```python
#4. Medidas estadisticas 
# Precio promedio por categorias de productos
promedio_categorias = df.groupby('Product Category')['Price'].mean()
print(promedio_categorias)

# Conteo de productos por categoría
products= print(df['Product Category'].value_counts())
```

```python
#5. Grafico
# Conteo de productos por categoría
plt.figure(figsize=(8, 4))
df['Product Category'].value_counts().plot(kind='bar', color='#2b5c8f')
plt.title('Cantidad de Productos Únicos por Categoría')
plt.xlabel('Categoría')
plt.ylabel('Número de Productos')
plt.tight_layout()
plt.show()
```
